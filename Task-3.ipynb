```python
#1.Employee Performance Bonus Eligibility

employees={
    "Ravi":92,
    "Anita":88,
    "Kiran":92,
    "Suresh":85
}
#step 1:Find highest score
highest_score=max(employees.values())

#step 2:Find all employees with highest score
top_performers=[name for name,score in employees.items() if score==highest_score]

#step 3:Display result
print("Top Performers Eligible for Bonus:",",".join(top_performers),f"(Score:{highest_score})")
```

```python
#2: Search Query Keyword Analysis
import string

# input
text = "Buy mobile phone buy phone online"

# Step 1: Convert to lowercase
text = text.lower()

# Step 2: Remove punctuation
text = text.translate(str.maketrans('', '', string.punctuation))

# Step 3: Split into words
words = text.split()

# Step 4: Count frequency
frequency = {}

for word in words:
    frequency[word] = frequency.get(word, 0) + 1

# Step 5: Display only words with count > 1
result = {word: count for word, count in frequency.items() if count > 1}

print(result)
```

```python
#3: Sensor Data Validation

sensor_readings=[3,4,7,8,10,12,5]

valid_readings=[]

for hour,value in enumerate(sensor_readings):
  if value%2==0:
    valid_readings.append((hour,value))

print("Valid Sensor Readings(Hour,Value):")
print(valid_readings)
```

```python
# 4: Email Domain Usage Analysis

emails=[
     "ravi@gmail.com",
     "anita@yahoo.com",
     "kiran@gmail.com",
     "suresh@gmail.com",
     "meena@yahoo.com"
]

domain_count={}

for email in emails:
  domain=email.split("@")[1]
  domain_count[domain]=domain_count.get(domain,0)+1

total=len(emails)

for domain,count in domain_count.items():
  percentage=(count/total)*100
  print(f"{domain}:{percentage:.0f}%")
```

```python
#5: Sales Spike Detection

sales=[1200,1500,900,2200,1400,3000]

#step 1:Average sales
average=sum(sales)/len(sales)

#step 2:Threshold(30% above average)
threshold=average*1.25

#step 3:Detect spike days
for day,value in enumerate(sales,start=1):
  if value>threshold:
    print(f"Day{day}:{value}")
```
