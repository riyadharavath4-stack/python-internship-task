```python
#1. 1:Smart Parking Lot Management System
def smart_parking(capacity, logs):
    current_vehicles = 0
    peak_usage = 0

    for log in logs:
        if log == "IN":
            current_vehicles += 1
        elif log == "OUT":
            if current_vehicles > 0:
                current_vehicles -= 1

        if current_vehicles > peak_usage:
            peak_usage = current_vehicles

    print("Currently Parked Vehicles:", current_vehicles)
    print("Peak Parking Usage:", peak_usage)

    if current_vehicles > capacity:
        print("Parking Status: Exceeded Capacity!")
    else:
        print("Parking Status: Available")


capacity = 50
logs = ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"]

smart_parking(capacity, logs)
```

```python
#2:Online Food Delivery Time Estimator
def estimate_delivery_time(distance, traffic, weather):
    eta = distance * 5

    if traffic.lower() == "high":
        eta += 10
    elif traffic.lower() == "medium":
        eta += 5

    if weather.lower() == "rainy":
        eta += 5
    elif weather.lower() == "stormy":
        eta += 15

    print("Estimated Delivery Time:", eta, "minutes")


distance = 8
traffic = "High"
weather = "Rainy"

estimate_delivery_time(distance, traffic, weather)
```

```python
#3.Movie Theatre Seat Occupancy Analyzer
def analyze_seat_occupancy(total_seats, booked_seats):
    booked_count = len(booked_seats)
    occupancy = (booked_count / total_seats) * 100

    print("Occupancy:", str(int(occupancy)) + "%")

    if occupancy == 100:
        print("Show Status: Housefull")
        print("Suggestion: Open additional show")
    elif occupancy >= 75:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Seats Available")


total_seats = 200
booked_seats = [1] * 150

analyze_seat_occupancy(total_seats, booked_seats)
```

```python
#4.Cloud Server Load Classification System
def classify_server_load(cpu_readings):
    if not cpu_readings:
        print("No CPU readings provided.")
        return

    total = 0
    for reading in cpu_readings:
        total += reading

    average = total / len(cpu_readings)

    if average < 50:
        status = "Normal"
    elif 50 <= average < 80:
        status = "Warning"
    else:
        status = "Critical"

    print(f"Average CPU Load: {average:.0f}%")
    print(f"Server Status: {status}")


cpu_data = [45, 60, 70, 85, 90]
classify_server_load(cpu_data)
```

```python
#5.Smart Classroom Resource Usage Monitor
def monitor_resource_usage(resource_usage):
    overused = []

    for resource, hours in resource_usage.items():
        if hours > 8:
            overused.append(resource)

    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")


usage_data = {
    "Projector": 6,
    "AC": 9,
    "Lights": 4
}

monitor_resource_usage(usage_data)
```

```python
#6.Online Event Registration Capacity Controller
def event_registration_controller(capacity, registrations):
    confirmed = 0
    waitlisted = 0

    for i in range(registrations):
        if confirmed < capacity:
            confirmed += 1
        else:
            waitlisted += 1

    if confirmed >= capacity:
        status = "Closed"
    else:
        status = "Open"

    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)


event_capacity = 100
total_registrations = 105

event_registration_controller(event_capacity, total_registrations)
```
