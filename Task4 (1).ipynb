```python
# 1.mobile recharge validation system

def validate_recharge():
  valid_plans = [199,299,399,599]

  while True:
    amount = int(input("Enter recharge amount: "))

    if amount < 50:
      print("Recharge amount must be >= 50")
    elif amount not in valid_plans:
      print("Invalid plan!Available plans:",valid_plans)
    else:
      print("Recharge successful!")
      break
validate_recharge()
```

```python
# 2.Inventory reorder alert system

def check_inventory(products):
  for item,stock in products.items():
    if stock < 15:
      print(f"{item}:Reorder Alert!")
    else:
      print(f"{item}:Stock Ok")
products = {
    "Rice":10,
    "oil":25,
    "Sugar":8,
    "Wheat":20
}

check_inventory(products)
```

```python
# 3.student result processing system

def calculate_result(marks):
  total = 0

  for mark in marks:
    total += mark
  average = total/len(marks)

  if average >= 50:
    print("Pass")
  else:
    print("Fail")
  return average
marks = [60,55,70,45]
avg = calculate_result(marks)
print("Average:",avg)
```

```python
#4.cab fare estimator with retry option

def calculate_fare(distance,peak_hour):
  fare = 50 + (distance * 12)

  if peak_hour.lower() == 'yes':
    fare += fare*0.25

  return fare

while True:
  distance = float(input("Enter distance in km:"))
  peak_hour = input("Is it peak hour?(yes/no):")

  total_fare = calculate_fare(distance,peak_hour)
  print("Total Fare = rupees{total_fare:2f}")

  choice = input("Do you want to calculate again?(yes/no):")

  if choice.lower() !='yes':
    break
```

```python
#5.employee attendance eligibility checker

def calculate_attendance(attendance):
  total_days = len(attendance)
  present_days = attendance.upper().count('p')

  percentage = (present_days/total_days)*100
  return percentage

  attendance = input("Enter attendance record(p/A):")

  percentage = calculate_attendance(attendance)
  print(f"Attendance Percentage:{percentage:2f}%")

  if percentage >= 75:
    print("Eligible")
  else:
    print("Not Eligible")
```

```python
#6.password strength checker

def check_password(password):
  if len(password) <8:
    return False

  has_digit = any(char.isdigit()for char in password)
  has_special = any(char in "@#$" for char in password)

  return has_digit and has_special

  password = input("Enter password:")

  if check_password(password):
    print("Strong Password")
  else:
    print("Weak Password")
```
