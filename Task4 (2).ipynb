```python
# 1.social media-post engagement analyzer
def analyze_post_engagement(likes_list):
  total_likes = 0
  for likes in likes_list:
    total_likes += likes
  print("Total Likes:",total_likes)
  if total_likes >= 1000:
      print("Post Status:Virus Post")
  else:
    print("Post Status:Normal Engagement")
likes_data = [200,150,300,400]
analyze_post_engagement(likes_data)
```

```python
# 3.Agriculture-Rainfall Adequacy Checker
def check_rainfall(rainfall_list, required_level):
    total = 0

    # Using loop to calculate total rainfall
    for rain in rainfall_list:
        total += rain

    # Calculating average rainfall
    average = total / len(rainfall_list)

    print("Average Rainfall:", int(average))

    # Conditional logic
    if average >= required_level:
        print("Rainfall Status: Adequate Rainfall")
    else:
        print("Rainfall Status: Inadequate Rainfall")
# Example usage
rainfall_data = [60, 75, 80, 73]   # Sample seasonal rainfall values
required = 70

check_rainfall(rainfall_data, required)
```

```python
#4.Social Media – Duplicate Account Detection Problem Statement
def detect_duplicate_accounts(usernames):
    unique_users = set(usernames)

    if len(unique_users) != len(usernames):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")


# Test Cases
users1 = ["john123", "alice45", "mark89", "john123"]
users2 = ["rahul01", "sneha22", "arjun33"]

detect_duplicate_accounts(users1)  # Duplicate exists
detect_duplicate_accounts(users2)  # No duplicate
```

```python
#7: System Monitoring – Application Health Checker
def check_system_health(errors):
    print(f"Error Count: {errors}")
    if errors == 0:
        status = "Healthy"
    elif errors <= 5:
        status = "Minor Issues"
    else:
        status = "Critical Issues"

    print(f"System Status: {status}")
# Test Case
check_system_health(7)
```

```python
#9: E-Learning – Student Attendance Eligibility System
def check_exam_eligibility(attendance_list):
    total_classes = len(attendance_list)
    attended = 0

    for status in attendance_list:
        if status == "Present":
            attended += 1

    percentage = (attended / total_classes) * 100
    print(f"Attendance Percentage: {percentage}")
    if percentage >= 75:
        print("Exam Eligibility: Eligible")
    else:
        print("Exam Eligibility: Not Eligible")
# Test Case
attendance = ["Present", "Present", "Absent", "Present", "Present"]
check_exam_eligibility(attendance)
```

```python
#10: Smart Electricity Bill Analyzer
def calculate_electricity_bill(units):
    bill = 0

    # Billing Slabs
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)

    # Usage Classification
    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"

    return bill, usage


# Test Case
units = 250
bill, status = calculate_electricity_bill(units)

print(f"Units Consumed: {units}")
print(f"Total Bill: ₹{bill}")
print(f"Usage Classification: {status}")
```
